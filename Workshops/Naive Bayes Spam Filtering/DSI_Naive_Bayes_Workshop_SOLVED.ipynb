# Introduction to Naive Bayes Classifiers


Naive Bayes Classfication, also shortened as NBC, is a type of *supervised machine learning* algorithm that is based on Bayes Theorem and conditional probabilities.

- **Supervised Machine Learning**: ML using labeled training data (i.e. when training a spam email filter, we use data that is emails that are labeled by people as either spam or not spam)

NBC has many applications but by far the most prominent one is in spam email filtering. In fact, Gmail uses a modified version of Naive Bayes in their spam filtering algorithm. Although the spam filtering algorithm that we will build today is not nearly as complex as the one used by Gmail, it's still a good introductory application to understanding NBC.

# The Math Behind The ML


## Introducing Bayes Formula

Recall back to Stats 1 (if you took it) you learned this formula, known as Bayes Theorem:

<br>

$$P(A | B)=\frac{P(A∩B)}{P(B)}=\frac{P(B|A)P(A)}{P(B)}$$

<br>

This formula is the basis for NBC. In NBC we have a class variable $C$ which represents the different classes we wish to discriminate against. For example, in spam email filtering, we are discriminating two classes, spam and not spam (which we refer to as an email being *ham*). This simply means we are deciding whether each email is spam, which we will represent with the variable $S$, or ham, which we will represent with the variable $H$. Our end goal is to find $P(S|\mathbf{x})$ and $P(H|\mathbf{x})$ where $\mathbf{x}=[x_1,x_2,…,x_n]$ is a vector of boolean values $x_1,x_2,…$ etc. representing whether we found certain word(s) $W$ in the email. This can basically be understood as asking the following question:

<br>

**"Given whether or not the words . . . (for example, 'money,' 'house,' 'password,' etc.) appear in this email, what is the probability this email is spam and what is the probability this email is not spam?"**

<br>


And this is where Bayes Theorem comes in! We can use it, along with probabilties obtained from training data, to calculate these values:

<br>

$$P(S|\mathbf{x})=\frac{P(S∩\mathbf{x})}{P(\mathbf{x})}=\frac{P(S,x_1,x_2,…,x_n)}{P(\mathbf{x})}$$

<br>

$$P(H|\mathbf{x})=\frac{P(H∩\mathbf{x})}{P(\mathbf{x})}=\frac{P(S,x_1,x_2,…,x_n)}{P(\mathbf{x})}$$

<br>

Note: The comma notation $P(S,x_1,x_2,…,x_n)$ just means the probability of the email being spam AND the word $x_1$, $x_2, …$ etc. appearing in the email. This is the same as intersection, just using different notation for joint probabilities.

Our end goal is to compare the values for $P(S|\mathbf{x})$ and $P(H|\mathbf{x})$ and see which is larger to decide whether or not the email is spam.

## Simplifying The Bayesian Formula

The formulas above are pretty difficult and tedious to calculate as-is, which is why NBC algorithms often use properties of probability to simplify the calculation process used to make our final classification (is the email spam or not spam?).

Let's look at our two formulas again:

<br>

$$P(S|\mathbf{x})=\frac{P(S,x_1,x_2,…,x_n)}{P(\mathbf{x})}$$

<br>

$$P(H|\mathbf{x})=\frac{P(H,x_1,x_2,…,x_n)}{P(\mathbf{x})}$$

<br>

Since our end goal is only to compare these two probabilities to see which is larger (we don't really care *what* these probabilities are, just which is larger), we can actually drop the denominator since it will be the same for both $P(S|\mathbf{x})$ and $P(H|\mathbf{x})$. This results in the following new relations:

<br>

$$P(S|\mathbf{x})\propto P(S,x_1,x_2,…,x_n)$$

<br>

$$P(H|\mathbf{x})\propto P(H,x_1,x_2,…,x_n)$$

<br>

The symbol $\propto$ just means "directly proportional to," which is true since we are treating $P(\mathbf{x})$ as a constant.

We can use the *chain rule for iterated conditional probability* to expand out the joint probabilities like so:

<br>

$$P(S,x_1,x_2,…,x_n)=P(x_1|x_2,\dots,x_n,S)P(x_2|x_3,\dots,x_n,S) \cdots P(x_{n-1}|x_n,S)P(x_n|S)P(S)$$

<br>

$$P(H,x_1,x_2,…,x_n)=P(x_1|x_2,\dots,x_n,H)P(x_2|x_3,\dots,x_n,H) \cdots P(x_{n-1}|x_n,H)P(x_n|H)P(H)$$

<br>

It's totally ok if you don't exactly understand where this expansion comes from, but it works quite similar to how the triple intersection formula does just repeated many times over:

<br>

$P(A∩B∩C)=P(A|B \cap C)P(B|C)P(C)$


## Why Is It "Naive" & Getting Our Final Bayesian Formula


The reason why Naive Bayes Classification is called "naive" is because it assumes *conditional independence* between each feature used for classification (in our case, the words of the email). As you'd suspect, this is not actually true to reality, as words in an email are far from independent of each other (like for example if we have the word "dog" in the email we would expect the next word to be something perhaps related to dogs like "shelter" or "adopt", etc). This is a common Bayesian practice to assume things we know to not exactly be true in the name of finding a *closed form solution*, or one that can be calculated using a finite number of (relatively simple) steps.

<br>

Conditional independence is defined as follows:

<br>

> It is said that $A$ is conditionally independent of $B$ given $C$ if and only if $P(A|B,C)=P(A|C)$.

<br>

Again, it alright if you don't understand exactly what this means or where it comes from, just know that we are doing this in the name of simplifying our final probability calculations. Since we assume conditional independence between the appearance of ALL words in the email, given they are from the same email, we can simplify the probabilities from before as follows:

<br>

- $P(x_1|x_2,\dots,x_n,S)=P(x_1|S)$

- $P(x_2|x_3,\dots,x_n,S)=P(x_2|S)$

- $P(x_{n-1}|x_n,S)=P(x_{n-1}|S)$

<br>

Likewise,

- $P(x_1|x_2,\dots,x_n,H)=P(x_1|H)$

- $P(x_2|x_3,\dots,x_n,H)=P(x_2|H)$

- $P(x_{n-1}|x_n,H)=P(x_{n-1}|H)$

<br>

Substituting these (much simpler) probabilities back into our original formula, we obtain:

<br>

$$P(S|\mathbf{x})\propto P(S,x_1,x_2,…,x_n)=P(x_1|S)P(x_2|S) \cdots P(x_{n-1}|S)P(x_n|S)P(S)$$

<br>

$$P(H|\mathbf{x})\propto P(H,x_1,x_2,…,x_n)=P(x_1|H)P(x_2|H) \cdots P(x_{n-1}|H)P(x_n|H)P(H)$$

<br>

This is the final formula we will use to calculate our probabilities and classify the emails as spam and ham. The next section goes over how to find these right-hand side probabilities and make our final spam filter.

# Naive Bayes Step-By-Step

The first step of any NBC algorithm is to first find the probabilities that will be used to make our final calculations for $P(S|\mathbf{x})$ and $P(H|\mathbf{x})$. These probabilities are just the ones from the simplified formula above:

<br>

- $P(x_1|S)$

- $P(x_2|S)$

    $\dots$

- $P(x_{n}|S)$

- $P(S)$

<br>

and

<br>

- $P(x_1|H)$

- $P(x_2|H)$

    $\dots$

- $P(x_{n}|H)$

- $P(H)$

<br>

Each word of interest $W$ (with corresponding boolean value of $x_1,x_2,…$ etc. representing whether we found that word in the email) is obtained by creating a vocabulary of unique words that appear among the entire training data set. The probabilities above are just the probability of observing each word in the vocabulary in an email given the email is spam or ham. These probabilities are calculated empirically (using real-life data) from a training set of emails that are already classified by people to be spam or ham. The next sections goes over exactly how we do this.

## Calculating the Probabilities

The process for obtaining the conditional probabilities above are:

<br>

---

<br>

1. Iterate through each of the emails in the training set (both spam and ham emails) and tokenize them (split them up into words)

<br>

---

<br>

2. For each word encoutered, add it to a dictionary of words. We will call this the vocabulary.

<br>

---

<br>

3. Filter through the vocabulary to remove repeated words and words that are not very interesting and don't tell us much about whether an email containing them is spam or not. These can be pronouns (she, her, his), conjunctions (and, or, but), etc.

<br>

---

<br>

4. Now that we have this vocabulary only with useful words, we will iterate through each word in the vocabulary and count how many spam emails in the training set contain that word as well as how many ham emails contain that word.

<br>

---

<br>

5. Now for each word $W$ in the vocabulary, we can calculate $P(x_{n}|S)$, or the probability we observe the word in an email given the email is spam using the following formula:

<br>

$$P(x_{n}|S)=\frac{1+N_{x_{n},S}}{2+N_{S}}$$

<br>

And we can calculate $P(x_{n}|H)$ similarly:

<br>

$$P(x_{n}|H)=\frac{1+N_{x_{n},H}}{2+N_{H}}$$

<br>

Where $N_{x_{n},S}$ / $N_{x_{n},H}$ is the number of spam / ham emails in the training set that contain the word $W$ and $N_{S}$ / $N_{H}$ is the total number of spam / ham emails in the training set.

Notice these formulas are quite similar to the classic conditional probability formula with the exception that we are adding 1 and 2 to the numerator and denominator, respectively. This is known as *smoothing*. The reason we do this is because there could be a word that appears in the training ham emails that does not appear in the training spam emails or vice versa. This word would result in a 0 probability we observe that word in an email given it is ham or spam. This is because we add this word to the vocabulary, count it 0 times in either the spam / ham training set, and that would result in a numerator of 0 and a resulting 0 conditional probability. This would cause us to multiply all the other $P(x|S)$ or $P(x|H)$ probabilities by 0, resulting in a 0 probability an email is spam / ham any time it contains a word that was found only in the one category of emails in the training set.

We obviously do not want this, so we modify the conditional probability formula to prevent these 0 probabilities from arising.

<br>

---

<br>

6. Now that we have all our conditional probabilities calculated, all we need to do is find $P(S)$ and $P(H)$, or the marginal probabilities that an email is spam or ham in general. These are easy to find: just take the proportions of your training data set that are marked as spam and ham, respectively.

## Finding Whether an Email is Spam / Ham

Now that we have all our probabilities, all there's left to do is calculate the directly proportional values to $P(S|\mathbf{x})$ and $P(H|\mathbf{x})$ and compare which is greater. Remember that $\mathbf{x}=[x_1,x_2,…,x_n]$ is a vector of boolean values $x_1,x_2,…$ etc. representing whether we found certain word(s) $W$ in the email we are trying to classify.

We get this vector by tokenizing the email we are trying to classify (similar to how we tokenized the training data set) and keep only the **unique** (no duplicates!) words which appear in the vocabulary we got from our training data. The reason we do this is because these words don't give any information about whether or not the email in question is spam / ham since we didn't see it in our training data set (so we have nothing to base our knowledge of the spammy-ness / hammy-ness of this word on and no corresponding empirical conditional probability).

Now all there is left to do is multiply the corresponding probabilites (which we found before) for the words in the email of interest according to the following formula:

<br>

$$P(S|\mathbf{x})\propto P(x_1|S)P(x_2|S) \cdots P(x_{n-1}|S)P(x_n|S)P(S)$$

<br>

$$P(H|\mathbf{x})\propto P(x_1|H)P(x_2|H) \cdots P(x_{n-1}|H)P(x_n|H)P(H)$$

<br>

- If the value proportional to $P(S|\mathbf{x})$ is greater than that of $P(H|\mathbf{x})$, the email is spam.

- If the value proportional to $P(H|\mathbf{x})$ is greater than that of $P(S|\mathbf{x})$, the email is not spam (ham).

# Final Notes and Additional Resources

Although NBC is not the most sophisticated algorithm around, it is still a very useful machine learning paradigm for its relative simplicity and low computation costs that scale well with large training data sets. Like many ML algorithms, NBC becomes more accurate the more diverse training data that is input.  

Specifically, for large data sets, we often risk *underflow errors*, or when we try to compute and store a number too close to 0 that the computer simply rounds it down to 0. As you can imagine, when we multiply many, many small probabilities there comes a point when the computer is simply not precise enough to actually represent it.

To prevent this, we often use log-probabilities since logarithms convert very very small numbers to numbers that are larger in magnitude. You can read more about this practice and generally more on NBC [here](https://en.wikipedia.org/wiki/Naive_Bayes_classifier#Bernoulli_naive_Bayes:~:text=Often%2C%20it%20is%20only%20necessary%20to%20discriminate%20between%20classes.%20In%20that%20case%2C%20the%20scaling%20factor%20is%20irrelevant%2C%20and%20it%20is%20sufficient%20to%20calculate%20the%20log%2Dprobability%20up%20to%20a%20factor%3A).

NBC has many formulations that we did not go over today and I highly recommend reading [this article](https://digitalcommons.morris.umn.edu/cgi/viewcontent.cgi?article=1024&context=horizons) if you are interested in learning more about NBC and its various optimizations!
